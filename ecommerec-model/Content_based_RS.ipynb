# **Build Recommendation sysetm model**

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
data = pd.read_csv('amazon_products.csv')
data.head(12)
```

```python
# Calculate TF-IDF matrix and cosine similarity matrix globally
product_features = data['product_name'] + ' ' + data['about_product'] + ' ' + data['main_category'] + ' ' + data['sub_category']
tfidf_vectorizer = TfidfVectorizer(max_features=500, stop_words='english', lowercase=True)
tfidf_matrix = tfidf_vectorizer.fit_transform(product_features)
similarity_matrix = cosine_similarity(tfidf_matrix, tfidf_matrix)
```

```python
def get_similar_products(data, product_name, similarity_matrix, top_k=5):

  product=data[data['product_name'] == product_name].index[0]

  similer_product_score=list(enumerate(similarity_matrix[product]))

  similer_product_score=sorted(similer_product_score,key=lambda x:x[1],reverse=True)

  similarity_scores=similer_product_score[1:top_k+1]

  similarity_list = []
  for index, score in similarity_scores:
      product_info = data.iloc[index]
      similarity_list.append({
          'product_name': product_info['product_name'],
          'sub_category': product_info['sub_category'],
          'score': score
      })

  return pd.DataFrame(similarity_list)
```

```python
from collections import defaultdict

interaction_types = ['view', 'add_to_cart']
weights = {'view': 1.0, 'add_to_cart': 1.5}

# Example user history: (user_id, product_id, interaction_type)
user_interction_history = [
    (1, 'B09YLX91QR', 'view'), #cable
    (1, 'B07232M876', 'view'), #cable
    (1, 'B09LJ116B5', 'add_to_cart'), #Smartphones

    (2, 'B00N1U7JXM', 'view'), #paper
    (2, 'B09LJ116B5', 'add_to_cart'), #Smartphones

    (3, 'B09PL79D2X', 'view'), #Headphones
    (3, 'B09YLX91QR', 'add_to_cart'), #cable
    (3, 'B07232M876', 'add_to_cart'), #cable
]

def personalize_recommendations(user_id, user_history, data, similarity_matrix, top_k=5):
    user_interction_history = [n for n in user_history if n[0] == user_id]

    weights_by_product = defaultdict(float)
    interacted_product_ids = set()

    # Create a mapping from product_id to DataFrame index for efficient lookup
    product_id_to_index = {pid: idx for idx, pid in enumerate(data['product_id'])}

    for _, pid, interaction in user_interction_history:
        interacted_product_ids.add(pid)
        weights_by_product[pid] += weights.get(interaction, 1.0)

    score_aggregate = defaultdict(float)
    for pid, w in weights_by_product.items():
        if pid not in product_id_to_index:
            continue

        product_idx = product_id_to_index[pid]

        # Get similarity scores for the current product from the global matrix
        product_similarities = list(enumerate(similarity_matrix[product_idx]))

        # Sort by similarity score in descending order
        sorted_similarities = sorted(product_similarities, key=lambda x: x[1], reverse=True)

        # Iterate through similar products to aggregate scores
        for sim_idx, sim_score in sorted_similarities:
            recommended_product_id = data.iloc[sim_idx]['product_id']
            # Exclude the product itself and already interacted products
            if recommended_product_id == pid or recommended_product_id in interacted_product_ids:
                continue

            score_aggregate[recommended_product_id] += sim_score * w

    sorted_recs = sorted(score_aggregate.items(), key=lambda x: x[1], reverse=True)

    # Prepare output: map product_id back to product_name and score
    display_recommendation_list = []
    for rid, score in sorted_recs[:top_k]:
        product_info = data[data['product_id'] == rid].iloc[0]
        display_recommendation_list.append({'product_id': rid, 'sub_category': product_info['sub_category'], 'product_name': product_info['product_name'], 'score': float(score)})
    return display_recommendation_list

print('Personalized recommendations for user 1:')
display(pd.DataFrame(personalize_recommendations(1, user_interction_history, data, similarity_matrix, top_k=5)))
```
